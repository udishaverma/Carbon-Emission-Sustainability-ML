```python
# ============================================================
# CELL 1 - Data Loading & Basic Overview
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

DATA_PATH = "../data/raw/owid-co2-data.csv"
CODEBOOK_PATH = "../data/raw/owid-co2-codebook.csv"

df = pd.read_csv(DATA_PATH)
codebook = pd.read_csv(CODEBOOK_PATH)

print("=" * 60)
print("DATA LOADING & BASIC OVERVIEW")
print("=" * 60)
print("Dataset shape:", df.shape)
print("Codebook shape:", codebook.shape)
print("Rows:", df.shape[0], "| Columns:", df.shape[1])
print("Unique locations:", df["country"].nunique())
print("Unique years:", df["year"].nunique())

print("\nFirst 5 rows:")
display(df.head())
```

```python
# ============================================================
# CELL 2 - Dataset Structure & Codebook
# ============================================================

print("=" * 60)
print("DATASET STRUCTURE")
print("=" * 60)

print("\nData types:")
print(df.dtypes.value_counts())

print("\nDataset information:")
df.info()

print("\nCodebook columns:", list(codebook.columns))

print("\nCO₂ variable definition:")
display(codebook[codebook["column"] == "co2"])

print("\nFirst 10 codebook entries:")
display(codebook.head(10))
```

```python
# ============================================================
# CELL 3 - Geographic & Temporal Coverage
# ============================================================

print("=" * 60)
print("GEOGRAPHIC & TEMPORAL COVERAGE")
print("=" * 60)

print("Unique location labels:", df["country"].nunique())
print("Unique ISO codes:", df["iso_code"].nunique())
print("Missing ISO codes:", df["iso_code"].isna().sum())
print("Minimum year:", df["year"].min())
print("Maximum year:", df["year"].max())

records_per_year = df.groupby("year").size()
records_per_country = df.groupby("country").size()

print("\nRecords per year:")
display(records_per_year.head(10))

print("\nLocations with fewest records:")
display(records_per_country.sort_values().head(10))

print("\nLocations with most records:")
display(records_per_country.sort_values(ascending=False).head(10))

plt.figure(figsize=(12, 5))
plt.plot(records_per_year.index, records_per_year.values)
plt.xlabel("Year")
plt.ylabel("Number of Records")
plt.title("Number of Dataset Records by Year")
plt.grid(True)
plt.show()
```

```python
# ============================================================
# CELL 4 - Missing Values & Duplicate Analysis
# ============================================================

print("=" * 60)
print("MISSING VALUES & DUPLICATE ANALYSIS")
print("=" * 60)

missing_count = df.isna().sum()
missing_percentage = (df.isna().mean() * 100).round(2)

missing_summary = pd.DataFrame({"missing_count": missing_count, "missing_percentage": missing_percentage}).sort_values("missing_percentage", ascending=False)

print("Duplicate rows:", df.duplicated().sum())
print("Duplicate country-year combinations:", df.duplicated(subset=["country", "year"]).sum())

print("\nTop 20 variables by missing percentage:")
display(missing_summary.head(20))

print("\nMissing values in CO₂:", df["co2"].isna().sum())
print("CO₂ missing percentage:", round(df["co2"].isna().mean() * 100, 2), "%")

top_missing = missing_summary.head(20)
plt.figure(figsize=(12, 7))
plt.barh(top_missing.index[::-1], top_missing["missing_percentage"][::-1])
plt.xlabel("Missing Values (%)")
plt.ylabel("Variable")
plt.title("Top 20 Variables by Missing-Value Percentage")
plt.show()
```

```python
# ============================================================
# CELL 5 - P_132 Variable Investigation
# ============================================================

print("=" * 60)
print("P_132 VARIABLE INVESTIGATION")
print("=" * 60)

keywords = ["co2", "energy", "renew", "coal", "oil", "gas", "gdp", "population"]

for keyword in keywords:
    matches = [col for col in df.columns if keyword in col.lower()]
    print(f"\n{keyword.upper()} variables:", matches)

candidate_columns = [
    "co2", "co2_per_capita", "co2_per_gdp", "population", "gdp",
    "primary_energy_consumption", "coal_co2", "oil_co2", "gas_co2",
    "cement_co2", "flaring_co2", "renewables", "renewables_share_energy",
    "renewables_share_elec"
]

available_candidates = [col for col in candidate_columns if col in df.columns]
missing_candidates = [col for col in candidate_columns if col not in df.columns]

print("\nAvailable candidate variables:", available_candidates)
print("Candidate variables not found:", missing_candidates)

candidate_summary = pd.DataFrame({
    "missing_count": df[available_candidates].isna().sum(),
    "missing_percentage": df[available_candidates].isna().mean().mul(100).round(2)
})

print("\nCandidate variable missing-value summary:")
display(candidate_summary.sort_values("missing_percentage", ascending=False))

print("\nCandidate variable statistics:")
display(df[available_candidates].describe().T)
```

```python
# ============================================================
# CELL 6 - Energy Dataset Loading & Basic Overview
# ============================================================

ENERGY_DATA_PATH = "../data/raw/owid-energy-data.csv"
ENERGY_CODEBOOK_PATH = "../data/raw/owid-energy-codebook.csv"

energy_df = pd.read_csv(ENERGY_DATA_PATH)
energy_codebook = pd.read_csv(ENERGY_CODEBOOK_PATH)

print("=" * 60)
print("ENERGY DATASET LOADING & BASIC OVERVIEW")
print("=" * 60)
print("Energy dataset shape:", energy_df.shape)
print("Energy codebook shape:", energy_codebook.shape)
print("Rows:", energy_df.shape[0], "| Columns:", energy_df.shape[1])
print("Unique locations:", energy_df["country"].nunique())
print("Unique years:", energy_df["year"].nunique())

print("\nFirst 5 rows:")
display(energy_df.head())
```

```python
# ============================================================
# CELL 7 - Energy Dataset Structure & Codebook
# ============================================================

print("=" * 60)
print("ENERGY DATASET STRUCTURE")
print("=" * 60)

print("\nData types:")
print(energy_df.dtypes.value_counts())

print("\nDataset information:")
energy_df.info()

print("\nCodebook columns:", list(energy_codebook.columns))

print("\nFirst 10 codebook entries:")
display(energy_codebook.head(10))
```

```python
# ============================================================
# CELL 8 - Energy Geographic & Temporal Coverage
# ============================================================

print("=" * 60)
print("ENERGY GEOGRAPHIC & TEMPORAL COVERAGE")
print("=" * 60)

print("Unique location labels:", energy_df["country"].nunique())
print("Unique ISO codes:", energy_df["iso_code"].nunique())
print("Missing ISO codes:", energy_df["iso_code"].isna().sum())
print("Minimum year:", energy_df["year"].min())
print("Maximum year:", energy_df["year"].max())

energy_records_per_year = energy_df.groupby("year").size()

print("\nFirst 10 yearly record counts:")
display(energy_records_per_year.head(10))

print("\nLast 10 yearly record counts:")
display(energy_records_per_year.tail(10))
```

```python
# ============================================================
# CELL 9 - Energy Missing Values & Duplicate Analysis
# ============================================================

print("=" * 60)
print("ENERGY MISSING VALUES & DUPLICATE ANALYSIS")
print("=" * 60)

energy_missing_count = energy_df.isna().sum()
energy_missing_percentage = (energy_df.isna().mean() * 100).round(2)

energy_missing_summary = pd.DataFrame({
    "missing_count": energy_missing_count,
    "missing_percentage": energy_missing_percentage
}).sort_values("missing_percentage", ascending=False)

print("Duplicate rows:", energy_df.duplicated().sum())
print("Duplicate country-year combinations:", energy_df.duplicated(subset=["country", "year"]).sum())

print("\nTop 20 variables by missing percentage:")
display(energy_missing_summary.head(20))

top_energy_missing = energy_missing_summary.head(20)

plt.figure(figsize=(12, 7))
plt.barh(top_energy_missing.index[::-1], top_energy_missing["missing_percentage"][::-1])
plt.xlabel("Missing Values (%)")
plt.ylabel("Variable")
plt.title("Top 20 Energy Variables by Missing-Value Percentage")
plt.show()
```

```python
# ============================================================
# CELL 10 - Renewable & Energy Variable Investigation
# ============================================================

print("=" * 60)
print("RENEWABLE & ENERGY VARIABLE INVESTIGATION")
print("=" * 60)

keywords = ["renew", "energy", "solar", "wind", "hydro", "biofuel", "fossil", "coal", "oil", "gas"]

for keyword in keywords:
    matches = [col for col in energy_df.columns if keyword in col.lower()]
    print(f"\n{keyword.upper()} variables:", matches)

renewable_columns = [col for col in energy_df.columns if "renew" in col.lower() or "biofuel" in col.lower()]
print("\nRenewable-related variables:")
print(renewable_columns)

print("\nRenewable variable definitions:")
display(energy_codebook[energy_codebook["column"].isin(renewable_columns)])
```

```python
# ============================================================
# CELL 11 - P_132 Candidate Variables in Both Datasets
# ============================================================

print("=" * 60)
print("P_132 VARIABLES IN BOTH DATASETS")
print("=" * 60)

candidate_keywords = ["co2", "energy", "renew", "coal", "oil", "gas", "solar", "wind", "hydro"]

co2_matches = sorted({col for keyword in candidate_keywords for col in df.columns if keyword in col.lower()})
energy_matches = sorted({col for keyword in candidate_keywords for col in energy_df.columns if keyword in col.lower()})

print("CO₂ dataset matching variables:")
print(co2_matches)

print("\nEnergy dataset matching variables:")
print(energy_matches)

print("\nCommon columns:")
print(sorted(set(df.columns) & set(energy_df.columns)))

print("\nPotential renewable-energy variables from Energy dataset:")
print(renewable_columns)
```
